Esturctura PARA-for
* La estructura PARA se emplea cuando se sabe ** cuantas veces** se repite una accion.

Sintaxis
´´´´´´´´
for (variable)** in range**(VALOR INICIAL,VALOR FINAL,pasos):
  La accion que se repite VF-VI


```python
#Hacer un programa que calcule la nomina de n empleados
nomina=0
n=int(input("Ingrese la cantidad de empleados:"))
for s in range(1,n+1,1):
  sueldo=int(input(f"Ingrese el salario del empleado{s}:"))
  nomina=nomina+sueldo #Acumulador
  print(f"La nomina de {n}empleados es: {nomina}")
```

```python
#Hacer un programa que calcule la nomina de n empleados
nomina=0
n=int(input("Ingrese la cantidad de empleados:"))
for s in range(n):
  sueldo=int(input(f"Ingrese el salario del empleado{s+1}:"))
  nomina=nomina+sueldo #Acumulador
  print(f"La nomina de {n}empleados es: {nomina}")
```
